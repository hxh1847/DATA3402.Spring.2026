```python
# 1 & 2 Matrix Class
class matrix:
    def __init__(self, n, m=None):
        # Case 1: matrix(n, m) -> zero matrix
        if isinstance(n, int) and isinstance(m, int):   # FIX 1: was 'isisyance'
            self.data = [[0.0 for _ in range(m)] for _ in range(n)]

        # Case 2: matrix(list_of_lists)
        elif isinstance(n, list):
            row_length = len(n[0])
            for row in n:
                if len(row) != row_length:
                    raise ValueError("All rows must have the same length.")
            self.data = [[float(value) for value in row] for row in n]

        else:
            raise ValueError("Invalid input for matrix initialization.")

    # Prints the matrix nicely instead of showing a memory address
    def __repr__(self):                                  # FIX 10: added __repr__
        rows_str = []
        for row in self.data:
            vals = [str(int(v)) if v == int(v) else str(v) for v in row]
            rows_str.append("[ " + "  ".join(vals) + " ]")
        return "\n".join(rows_str)

    # shape
    def shape(self):
        return (len(self.data), len(self.data[0]))

    # indexing: M[i][j] and M[i, j]
    def __getitem__(self, index):
        if isinstance(index, tuple):
            i, j = index            # FIX 2: i and j were never unpacked
            return self.data[i][j]
        return self.data[index]

    def __setitem__(self, index, value):
        i, j = index
        self.data[i][j] = float(value)

    # assignment: copies values from another matrix or list of lists
    def assign(self, other):
        src = other if isinstance(other, matrix) else matrix(other)
        if self.shape() != src.shape():
            raise ValueError("Shapes must match for assignment.")
        self.data = [row[:] for row in src.data]

    # to_list
    def to_list(self):
        return self.data

    # transpose
    def transpose(self):
        rows, cols = self.shape()
        result = matrix(cols, rows)
        for i in range(rows):
            for j in range(cols):
                result[j, i] = self[i, j]
        return result

    # row
    def row(self, n):
        return matrix([self.data[n]])

    # column
    def column(self, n):
        col = [[self.data[i][n]] for i in range(len(self.data))]
        return matrix(col)          # FIX 3: was missing return

    # block
    def block(self, n0, n1, m0, m1):
        block_data = []
        for i in range(n0, n1):     # FIX 4: was range(n0, m1), should be n1
            row = []
            for j in range(m0, m1):
                row.append(self[i, j])
            block_data.append(row)
        return matrix(block_data)

    # scalar multiplication
    def scalarmul(self, c):
        rows, cols = self.shape()
        result = matrix(rows, cols)
        for i in range(rows):
            for j in range(cols):
                result[i, j] = self[i, j] * c
        return result

    # addition
    def add(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrices must be same size for addition.")
        rows, cols = self.shape()
        result = matrix(rows, cols) # FIX 5: was 'results', returned 'result'
        for i in range(rows):
            for j in range(cols):
                result[i, j] = self[i, j] + N[i, j]
        return result

    # subtraction
    def sub(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrices must be same size for subtraction.")
        rows, cols = self.shape()
        result = matrix(rows, cols) # FIX 6: was 'results', returned 'result'
        for i in range(rows):
            for j in range(cols):
                result[i, j] = self[i, j] - N[i, j]
        return result

    # matrix multiplication
    def mat_multi(self, N):
        r1, c1 = self.shape()
        r2, c2 = N.shape()
        if c1 != r2:
            raise ValueError("Invalid dimensions for matrix multiplication.")
        result = matrix(r1, c2)
        for i in range(r1):
            for j in range(c2):
                total = 0
                for k in range(c1):
                    total += self[i, k] * N[k, j]
                result[i, j] = total
        return result

    # element-wise multiplication
    def element_mult(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrices must be same size.")
        rows, cols = self.shape()
        result = matrix(rows, cols) # FIX 7: was 'results', returned 'result'
        for i in range(rows):
            for j in range(cols):
                result[i, j] = self[i, j] * N[i, j]
        return result

    # equality
    def equals(self, N):
        return self.data == N.data

    # Operator overloading
    def __add__(self, N):    return self.add(N)
    def __sub__(self, N):    return self.sub(N)
    def __mul__(self, other):
        if isinstance(other, matrix):
            return self.mat_multi(other) # FIX 8: was self.mat_mult (wrong name)
        else:
            return self.scalarmul(other)
    def __rmul__(self, c):   return self.scalarmul(c)  # FIX 9: added for 2*M
    def __eq__(self, N):     return self.equals(N)


print("Matrix class ready!")
```

```python
# 3.Special Matrices 

def constant(n, m, c):
    return matrix([[float(c) for _ in range(m)] for _ in range(n)])

def zeros(n, m):
    return constant(n, m, 0.0)

def ones(n, m):
    return constant(n, m, 1.0)

def eye(n):
    I = zeros(n, n)
    for i in range(n):
        I[i, i] = 1.0
    return I

print("constant(2,3,7):"); print(constant(2, 3, 7))
print("\nzeros(2,3):");    print(zeros(2, 3))
print("\nones(2,3):");     print(ones(2, 3))
print("\neye(3):");        print(eye(3))
```

```python
# 6. matrix Properties
A = matrix([[1, 2],
            [3, 4]])
B = matrix([[2, 0],
            [1, 2]])
C = matrix([[0, 1],
            [1, 0]])
I = eye(2)

# Associative: (AB)C = A(BC)
left  = (A * B) * C
right = A * (B * C)
print("Associative (AB)C == A(BC):", left == right)

# Distributive: A(B+C) = AB + AC
left  = A * (B + C)
right = (A * B) + (A * C)
print("Distributive A(B+C) == AB+AC:", left == right)

# Non-commutative: AB != BA
print("AB == BA (should be False):", (A * B) == (B * A))

# Identity: AI = A
print("AI == A:", (A * I) == A)
```
