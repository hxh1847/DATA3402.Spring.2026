```python
# Exercise 2: Write a function that takes two integers n and m and draws an n by m game board.
def draw_board(n, m):
    print(' ---' * m)
    
    for i in range(n):
        print('|  ' * m + '|')
        print(' ---' * m)
draw_board(3, 3)
```

```python
# Exercise 3: Modify Exercise 2 so that it takes a matrix in the format from Exercise 1 and draws a tic‑tac‑toe board with "X"s and "O"s.
def draw_board(board):
    n = len(board) #number of rows
    m = len(board[0]) #number of columns

    # symbol mapping
    symbols = {0: ' ', 1: 'X', 2: 'O'}

    # top border
    print(' ---' * m)

    #for each row
    for row in board:
        # buidling cell content line
        line = '|'
        for cell in row:
            line += ' ' + symbols[cell] + ' |'
        print(line)
        print(' ---' * m)
test_board = [[1, 0, 2],
              [2, 1, 0],
              [0, 0, 1]]
draw_board(test_board)
```

```python
# Exercise 4: Write a function that takes an n×n matrix representing a tic‑tac‑toe game and returns one of the following values:
# -1 if the game is incomplete (still empty spaces and no winner)
#0 if the game is a draw
#1 if player 1 ("X") has won
#2 if player 2 ("O") has won

def check_winner(board):
    n = len(board)

    # checking rows (horizontal)
    for row in board:
        if all(cell == 1 for cell in row): #if all cells = 1
            return 1 # returning loop 
        if all(cell == 2 for cell in row):
            return 2
    # checking columns (vertical)
    for col in range(n):
        if all(board[row][col] == 1 for row in range(n)):
            return 1
        if all(board[row][col] == 2 for row in range(n)):
            return 2

    # checking main diagonal (top left to bottom right)
    if all(board[i][i] == 1 for i in range(n)):
        return 1
    if all(board[i][i] == 2 for i in range(n)):
        return 2

    #checking anti-diagonal (top-right to bottom left)
    if all(board[i][n-1-i] == 1 for i in range(n)):
        return 1
    if all(board[i][n-1-i] == 2 for i in range(n)):
        return 2

    # checking if game is complete or still going 
    empty = any(0 in row for row in board)
    if empty:
        return -1
    else:
        return 0
board = [[1, 2, 0], [2, 1, 0], [2, 1, 1]]
print(check_winner(board))
```

```python
# Exercise 5: Write a function that takes a game board, a player number, and (row, col) coordinates and places the correct mark ("X" or "O") in that location.

def place_mark(board, player, row, col):
    if board[row][col] == 0:
        board[row][col] = player
        return True
    else:
        return False
board = create_board(3)
print(place_mark(board, 1, 0, 0))
print(board)
print(place_mark(board, 2, 0, 0))

board = create_board(3)
place_mark(board, 1, 0, 0)
print(board[0][0])
```

```python
# Exercise 6: Modify Exercise 3 to show row and column labels so that players can specify locations like "A2" or "C1".

def draw_labeled_board(board):
    n = len(board)
    symbols = {0: ' ', 1: 'X', 2: 'O'}
    
    # Column headers (A, B, C, ...)
    header = '   '  # Indent for row numbers
    for i in range(n):
        header += ' ' + chr(65 + i) + '  '  # chr(65) = 'A'
    print(header)
    
    # Top border
    print('  ' + ' ---' * n)
    
    # Rows with numbers
    for idx, row in enumerate(board):
        # Row number
        line = str(idx + 1) + ' |'
        for cell in row:
            line += ' ' + symbols[cell] + ' |'
        print(line)
        print('  ' + ' ---' * n)
        
board = [[1, 0, 2], [2, 1, 0], [0, 0, 1]]
draw_labeled_board(board)
```

```python
# Exercise 7: Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

def parse_and_place(board, player, location):
    """
    Convert 'A2' to row/col and place mark.
    Returns True if successful.
    """
    # getting letter and number
    letter = location[0].upper()  # 'A', 'B', 'C'
    number = location[1]           # '1', '2', '3'
    
    # converting to indices
    col = ord(letter) - ord('A')  # 'A'->0, 'B'->1, 'C'->2
    row = int(number) - 1          # '1'->0, '2'->1, '3'->2
    
    # checking if valid
    n = len(board)
    if row < 0 or row >= n or col < 0 or col >= n:
        return False
    
    return place_mark(board, player, row, col)

# testing it
board = create_board(3)
parse_and_place(board, 1, 'A1')  # X at top-left
parse_and_place(board, 2, 'B2')  # O at center
draw_labeled_board(board)
```

```python
# Exercise 8: Write a function that is called with a board and player number, takes input from the player using Python's input(), and modifies the board using your function from Exercise 7.

def get_player_move(board, player):
    """Ask player for input until they give a valid move"""
    
    if player == 1:
        symbol = 'X'
    else:
        symbol = 'O'
    
    while True:
        location = input(f"Player {player} ({symbol}), enter move (like A1): ")
        
        success = parse_and_place(board, player, location)
        
        if success:
            print("Good move!")
            break
        else:
            print("Invalid move. Try again.")


# Test 1: Manual testing (run this interactively)
board = create_board(3)
print("Testing get_player_move - Board before:")
draw_labeled_board(board)

print("\nTry entering 'A1' when prompted:")
get_player_move(board, 1)

print("\nBoard after move:")
draw_labeled_board(board)

print("\nTry entering 'A1' again (should reject it):")
get_player_move(board, 2)

print("\nTry entering 'B2' when prompted:")
get_player_move(board, 2)

print("\nFinal board:")
draw_labeled_board(board)
```

```python
# Exercise 9: Use all of the previous exercises to implement a full tic‑tac‑toe game:

def play_game(n=3):
    """Play a complete game"""
    
    # setup
    board = create_board(n)
    current_player = 1
    
    print(f"Let's play {n}×{n} Tic-Tac-Toe!")
    print("Player 1 = X, Player 2 = O")
    print()
    
    # game loop
    while True:
        # showing board
        draw_labeled_board(board)
        
        # get move
        get_player_move(board, current_player)
        
        # checking if someone won
        result = check_winner(board)
        
        if result == 1:
            draw_labeled_board(board)
            print("Player 1 (X) wins!")
            break
        elif result == 2:
            draw_labeled_board(board)
            print("Player 2 (O) wins!")
            break
        elif result == 0:
            draw_labeled_board(board)
            print("It's a draw!")
            break
        
        # switch players
        if current_player == 1:
            current_player = 2
        else:
            current_player = 1
        
        print() 
        
# test 1
print("=== Testing 3×3 Game ===")
play_game(3)
```

```python
# Exercise 11: Develop a version of the game where one player is the computer.
import random

def get_computer_move_simple(board, player):
    """
    Simple computer player strategy:
    1. Try to win if possible
    2. Block opponent if they're about to win
    3. Otherwise, pick a random empty spot
    """
    n = len(board)
    opponent = 2 if player == 1 else 1
    
    # try to find a winning move
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                # try move
                board[row][col] = player
                if check_winner(board) == player:
                    # winning move found! Keep it
                    print(f"Computer (Player {player}) moves to {chr(65+col)}{row+1}")
                    return
                # undo move
                board[row][col] = 0
    
    # trying to block opponent's winning move
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                # try opponent's move
                board[row][col] = opponent
                if check_winner(board) == opponent:
                    # block this winning move
                    board[row][col] = player
                    print(f"Computer (Player {player}) blocks at {chr(65+col)}{row+1}")
                    return
                # undo test
                board[row][col] = 0
    
    # no winning/blocking move, pick random empty spot
    empty_spots = []
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                empty_spots.append((row, col))
    
    if empty_spots:
        row, col = random.choice(empty_spots)
        board[row][col] = player
        print(f"Computer (Player {player}) moves to {chr(65+col)}{row+1}")

def play_game_vs_computer(n=3):
    """Play against the computer"""
    
    board = create_board(n)
    current_player = 1
    
    print(f"Let's play {n}×{n} Tic-Tac-Toe!")
    print("You are Player 1 (X)")
    print("Computer is Player 2 (O)")
    print()
    
    while True:
        draw_labeled_board(board)
        
        if current_player == 1:
            # human player
            get_player_move(board, current_player)
        else:
            # computer player
            get_computer_move_simple(board, current_player)
        
        result = check_winner(board)
        
        if result == 1:
            draw_labeled_board(board)
            print("🎉 You win! 🎉")
            break
        elif result == 2:
            draw_labeled_board(board)
            print("💻 Computer wins! 💻")
            break
        elif result == 0:
            draw_labeled_board(board)
            print("🤝 It's a draw! 🤝")
            break
        
        if current_player == 1:
            current_player = 2
        else:
            current_player = 1
        print()


# test 1: play against the simple computer
print("=== Playing vs Simple Computer ===")
play_game_vs_computer(3)
```
