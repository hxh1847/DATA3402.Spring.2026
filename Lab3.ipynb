```python
def generate_uniform(N, x_min, x_max):
    out = []
    for _ in range(N):
        r = random.random()               # r in [0,1)
        x = x_min + r * (x_max - x_min)   # scale + shift
        out.append(x)
    return out

# Test your solution here
data=generate_uniform(1000,-10,10)
print ("Data Type:", type(data))
print ("Data Length:", len(data))
if len(data)>0: 
    print ("Type of Data Contents:", type(data[0]))
    print ("Data Minimum:", min(data))
    print ("Data Maximum:", max(data))
```

```python
def mean(Data):
    total = 0.0
    for x in Data:
        total += x
    return total / len(Data)
print("Mean of Data:", mean(data))
```

```python
# Exercise 2b(3): Variance Fucntion
# Write a function that computes the variance of values in a list

def variance(Data):
    m = mean(Data)
    total = 0.0
    for x in Data:
        total += (x - m) **2
    return total / len(Data)
print("Variance of Data:", variance(data))
    
```

```python
# The list comprehension way
def variance(Data):
    m = mean(Data)

    return sum([(x - m)**2 for x in Data]) / len(Data)
```

```python
# Exercise 4:
def histogram(x, n_bins=10, x_min=None, x_max=None):
    if x_min is None:
        x_min = min(x)
    if x_max is None:
        x_max = max(x)

    bin_width = (x_max - x_min) / n_bins
    hist = [0] * n_bins
    bin_edges = []

    for i in range(n_bins + 1):
        bin_edges.append(x_min + i * bin_width)

    for value in x:
        index = int((value - x_min) / bin_width)
        if index == n_bins:
            index -= 1
        hist[index] += 1

    return hist, bin_edges
# Testing solution   
# Test your solution here
data = generate_uniform(10000, 0, 10)

h, b = histogram(data, 10)

print("Histogram counts:", h)
print("Bin edges:", b)
print("Total counts:", sum(h))
```

```python
# Exercise 5:

def draw_histogram(x, n_bins, x_min=None, x_max=None,
                   character="#", max_character_per_line=20):

    hist, bin_edges = histogram(x, n_bins, x_min, x_max)
    max_count = max(hist)

    for i in range(n_bins):
        bar_length = int(hist[i] / max_count * max_character_per_line)
        bar = character * bar_length
        print(f"[{bin_edges[i]:6.2f}, {bin_edges[i+1]:6.2f}] : {bar}")

    return hist, bin_edges
```

```python
# Exercise 8:
def generate_function(func, x_min, x_max, N=1000):
    out = []
    for i in range(N):
        x = x_min + i * (x_max - x_min) / (N - 1)
        out.append(func(x))
    return out
# Test your solution here
import math

values = generate_function(math.sin, 0, math.pi, 10)
print(values)
```

```python
# Exercise 8: 
def integrate(func, x_min, x_max, n_points=1000):
    dx = (x_max - x_min) / n_points
    total = 0.0
    for i in range(n_points):
        x = x_min + i * dx
        total += func(x) * dx
    return total
    
# Testing solution 
import math

result = integrate(math.sin, 0, math.pi, 10000)
print(result)
```
