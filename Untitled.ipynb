```python
class Card:
    pass

class Deck:
    pass

class Hand:
    pass

class Player:
    pass

class StrategyPlayer(Player):
    pass

class Game:
    pass

```

```python
# 1
class Card:

    def __init__(self, rank, suit):
        self.rank = rank
        self.suit = suit

    def value(self):

        if self.rank in ['J','Q','K']:
            return 10
        elif self.rank == 'A':
            return 11
        else:
            return int(self.rank)

    def __str__(self):
        return f"{self.rank} of {self.suit}"
```

```python
class Hand:

    def __init__(self):
        self.cards = []

    def add_card(self, card):
        self.cards.append(card)

    def total(self):

        total = sum(card.value() for card in self.cards)

        aces = sum(1 for c in self.cards if c.rank == 'A')

        while total > 21 and aces:
            total -= 10
            aces -= 1

        return total

    def show(self):
        return [str(card) for card in self.cards]
```

```python
class Player:

    def __init__(self, name, chips=100):
        self.name = name
        self.chips = chips
        self.hand = Hand()

    def reset_hand(self):
        self.hand = Hand()

    def play(self, deck):

        while self.hand.total() < 17:
            self.hand.add_card(deck.draw())
```

```python
class StrategyPlayer(Player):

    def __init__(self, name, threshold=0, chips=100):

        super().__init__(name, chips)

        self.running_count = 0
        self.threshold = threshold

    def update_count(self, card):

        if card.rank in ['2','3','4','5','6']:
            self.running_count += 1

        elif card.rank in ['10','J','Q','K','A']:
            self.running_count -= 1

    def play(self, deck):

        while self.running_count < self.threshold and self.hand.total() < 21:

            card = deck.draw()
            self.hand.add_card(card)
            self.update_count(card)
```

```python
class Game:

    def __init__(self, players):

        self.players = players
        self.dealer = Player("Dealer")
        self.deck = Deck()

    def deal(self):

        for player in self.players + [self.dealer]:
            player.reset_hand()

        for _ in range(2):

            for player in self.players + [self.dealer]:

                card = self.deck.draw()
                player.hand.add_card(card)

                for p in self.players:
                    if isinstance(p, StrategyPlayer):
                        p.update_count(card)

    def play_round(self):

        self.deal()

        for player in self.players:
            player.play(self.deck)

        self.dealer.play(self.deck)

        dealer_total = self.dealer.hand.total()

        for player in self.players:

            total = player.hand.total()

            if total > 21:
                player.chips -= 1

            elif dealer_total > 21 or total > dealer_total:
                player.chips += 1

            elif total < dealer_total:
                player.chips -= 1
```

```python
players = [
    StrategyPlayer("Counter", threshold=0),
    Player("Player1"),
    Player("Player2"),
    Player("Player3")
]

game = Game(players)

for i in range(10):

    game.play_round()

    print("Round", i+1)

    for p in players:
        print(p.name, "chips:", p.chips)

    print()
```

```python
# running 50 rounds 

def run_game(threshold=0):

    players = [
        StrategyPlayer("Counter", threshold),
        Player("P1"),
        Player("P2"),
        Player("P3")
    ]

    game = Game(players)

    for _ in range(50):

        game.play_round()

        if players[0].chips <= 0:
            break

    return players[0].chips
```

```python
results = []

for i in range(100):
    results.append(run_game(threshold=0))
```

```python
import matplotlib.pyplot as plt

plt.hist(results)
plt.title("Strategy Player Winnings")
plt.xlabel("Chips after 50 rounds")
plt.ylabel("Frequency")
plt.show()
```
